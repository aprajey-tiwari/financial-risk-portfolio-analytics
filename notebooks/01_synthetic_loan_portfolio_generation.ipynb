```python
# Import libraries required for synthetic data generation
import pandas as pd
import numpy as np

# Set a fixed random seed to make the output reproducible
rng = np.random.default_rng(42)

# Number of customers to generate
TOTAL_CUSTOMERS = 4000

print("Libraries imported successfully.")
print("Customers to generate:", TOTAL_CUSTOMERS)
```

```python
# Generate fictional customer information

first_names = [
    "Amit", "Rahul", "Neha", "Priya", "Ankit",
    "Rohit", "Sneha", "Karan", "Pooja", "Vikas",
    "Anjali", "Suresh", "Meera", "Arjun", "Kavita"
]

last_names = [
    "Sharma", "Verma", "Mehta", "Gupta", "Singh",
    "Patel", "Jain", "Agarwal", "Mishra", "Yadav"
]

locations = [
    ("Indore", "Madhya Pradesh"),
    ("Bhopal", "Madhya Pradesh"),
    ("Delhi", "Delhi"),
    ("Mumbai", "Maharashtra"),
    ("Pune", "Maharashtra"),
    ("Jaipur", "Rajasthan"),
    ("Lucknow", "Uttar Pradesh"),
    ("Bengaluru", "Karnataka"),
    ("Hyderabad", "Telangana"),
    ("Ahmedabad", "Gujarat")
]

# Generate unique customer IDs
customer_ids = [
    f"C{i:05d}" for i in range(1, TOTAL_CUSTOMERS + 1)
]

# Generate fictional customer names
customer_names = [
    f"{rng.choice(first_names)} {rng.choice(last_names)}"
    for _ in range(TOTAL_CUSTOMERS)
]

# Assign customer types
customer_types = rng.choice(
    ["Individual", "Business"],
    size=TOTAL_CUSTOMERS,
    p=[0.80, 0.20]
)

# Assign city and state together
location_indices = rng.integers(
    0, len(locations), size=TOTAL_CUSTOMERS
)

cities = [locations[i][0] for i in location_indices]
states = [locations[i][1] for i in location_indices]

# Generate onboarding dates
start_date = np.datetime64("2022-01-01")
end_date = np.datetime64("2026-10-31")

date_range = int((end_date - start_date) / np.timedelta64(1, "D"))

onboarding_dates = (
    start_date
    + rng.integers(0, date_range + 1, size=TOTAL_CUSTOMERS)
    .astype("timedelta64[D]")
)

# Combine everything into one customer table
customers = pd.DataFrame({
    "customer_id": customer_ids,
    "customer_name": customer_names,
    "customer_type": customer_types,
    "city": cities,
    "state": states,
    "onboarding_date": onboarding_dates
})

print("Customer dataset generated successfully.")
print("Total customer records:", len(customers))

customers.head(10)
```

```python

loan_details = loan_accounts.merge(
    product_reference[
        ["product_id", "interest_rate"]
    ],
    on="product_id",
    how="left"
)

loan_details.head(5)
```

```python

# Step 1: Convert annual interest rate into monthly interest rate
monthly_rate = loan_details["interest_rate"] / 100 / 12

# Step 2: Get loan tenure in months
n = loan_details["loan_tenure_months"]

# Step 3: Get original loan amount
P = loan_details["sanction_amount"]

# Step 4: Calculate monthly EMI
loan_details["monthly_emi"] = (
    P * monthly_rate * (1 + monthly_rate) ** n
) / (
    (1 + monthly_rate) ** n - 1
)

# Step 5: Round EMI to 2 decimal places
loan_details["monthly_emi"] = loan_details["monthly_emi"].round(2)

# Display the result
loan_details[
    ["account_id", "sanction_amount", "interest_rate",
     "loan_tenure_months", "monthly_emi"]
].head(10)
```

```python

# TARGET: Calculate the unpaid amount remaining on each instalment

payment_records["instalment_outstanding"] = (
    payment_records["amount_due"]
    - payment_records["amount_paid"]
).clip(lower=0)

# Display a sample for review
payment_records[
    [
        "account_id",
        "instalment_number",
        "amount_due",
        "amount_paid",
        "instalment_outstanding",
        "payment_status"
    ]
].head(10)
```

```python

# TARGET: Calculate how many days unpaid instalments are overdue
# Reporting date for this portfolio: 31 October 2026

reporting_date = pd.Timestamp("2026-10-31")

# Initially assign 0 DPD to every instalment
payment_records["dpd"] = 0

# Identify instalments that are unpaid and past their due date
overdue_mask = (
    (payment_records["instalment_outstanding"] > 0)
    & (payment_records["due_date"] < reporting_date)
)

# Calculate DPD only for overdue instalments
payment_records.loc[overdue_mask, "dpd"] = (
    reporting_date - payment_records.loc[overdue_mask, "due_date"]
).dt.days

# Display the results
payment_records[
    [
        "account_id",
        "instalment_number",
        "due_date",
        "instalment_outstanding",
        "dpd"
    ]
].head(15)
```

```python

# TARGET: Classify each instalment into a delinquency bucket

conditions = [
    (payment_records["dpd"] == 0),
    (payment_records["dpd"].between(1, 30)),
    (payment_records["dpd"].between(31, 60)),
    (payment_records["dpd"].between(61, 90)),
    (payment_records["dpd"] > 90)
]

bucket_names = [
    "Current",
    "1-30 DPD",
    "31-60 DPD",
    "61-90 DPD",
    "90+ DPD"
]

payment_records["dpd_bucket"] = np.select(
    conditions,
    bucket_names,
    default="Review"
)

# Display a sample
payment_records[
    [
        "account_id",
        "instalment_number",
        "instalment_outstanding",
        "dpd",
        "dpd_bucket"
    ]
].head(15)
```

```python

# TARGET: Summarise outstanding exposure by DPD bucket

dpd_summary = (
    payment_records
    .groupby("dpd_bucket")
    .agg(
        total_instalments=("account_id", "count"),
        unique_accounts=("account_id", "nunique"),
        total_outstanding=("instalment_outstanding", "sum")
    )
    .reset_index()
)

# Calculate each bucket's share of total outstanding exposure
total_portfolio_outstanding = dpd_summary["total_outstanding"].sum()

dpd_summary["exposure_percentage"] = (
    dpd_summary["total_outstanding"]
    / total_portfolio_outstanding
    * 100
).round(2)

# Display the portfolio summary
dpd_summary
```

```python

# TARGET: Calculate overall delinquency indicators

# Identify instalments that are currently overdue
overdue_records = payment_records[
    (payment_records["dpd"] > 0) &
    (payment_records["instalment_outstanding"] > 0)
]

# Identify instalments with 30 or more DPD
dpd_30_plus = overdue_records[
    overdue_records["dpd"] >= 30
]

# Calculate portfolio indicators
total_accounts = payment_records["account_id"].nunique()

delinquent_accounts = overdue_records["account_id"].nunique()

overdue_amount = overdue_records["instalment_outstanding"].sum()

dpd_30_plus_amount = dpd_30_plus["instalment_outstanding"].sum()

total_instalment_outstanding = (
    payment_records["instalment_outstanding"].sum()
)

print("PORTFOLIO DELINQUENCY SUMMARY")
print("--------------------------------")
print("Total loan accounts:", total_accounts)
print("Delinquent loan accounts:", delinquent_accounts)
print("Total overdue instalment amount:", round(overdue_amount, 2))
print("30+ DPD instalment amount:", round(dpd_30_plus_amount, 2))
print("Total instalment outstanding:", round(total_instalment_outstanding, 2))
```

```python

# TARGET: Build a contractual amortisation schedule for one loan

sample_account_id = "L000001"

# Retrieve the loan details
loan = loan_details.loc[
    loan_details["account_id"] == sample_account_id
].iloc[0]

# Retrieve its scheduled instalments
schedule = (
    repayment_schedule.loc[
        repayment_schedule["account_id"] == sample_account_id
    ]
    .sort_values("instalment_number")
)

opening_balance = float(loan["sanction_amount"])
annual_rate = float(loan["interest_rate"])
monthly_emi = float(loan["monthly_emi"])

monthly_rate = annual_rate / 1200

amortisation_records = []

for _, instalment in schedule.iterrows():

    interest_component = round(
        opening_balance * monthly_rate, 2
    )

    # Adjust the final principal repayment for rounding
    principal_component = min(
        round(monthly_emi - interest_component, 2),
        opening_balance
    )

    closing_balance = round(
        opening_balance - principal_component, 2
    )

    amortisation_records.append({
        "account_id": sample_account_id,
        "instalment_number": instalment["instalment_number"],
        "due_date": instalment["due_date"],
        "opening_principal": round(opening_balance, 2),
        "emi": monthly_emi,
        "interest_component": interest_component,
        "principal_component": principal_component,
        "closing_principal": closing_balance
    })

    opening_balance = closing_balance

# Convert the schedule into a DataFrame
sample_amortisation = pd.DataFrame(amortisation_records)

display(sample_amortisation)

print(
    "\nFinal principal balance:",
    sample_amortisation["closing_principal"].iloc[-1]
)
```

```python

# TARGET: Calculate actual principal outstanding after each instalment

# Retrieve actual payment records for the sample loan
actual_payments = (
    payment_records.loc[
        payment_records["account_id"] == sample_account_id,
        ["instalment_number", "amount_paid", "payment_status", "payment_date"]
    ]
    .copy()
)

actual_payments["amount_paid"] = (
    actual_payments["amount_paid"].fillna(0)
)

# Match scheduled instalments with actual payment records
actual_schedule = (
    sample_amortisation[
        ["instalment_number", "due_date", "emi"]
    ]
    .merge(
        actual_payments,
        on="instalment_number",
        how="left",
        validate="one_to_one"
    )
    .sort_values("instalment_number")
)

opening_principal = float(loan["sanction_amount"])
monthly_rate = float(loan["interest_rate"]) / 1200

actual_records = []

for _, row in actual_schedule.iterrows():

    # Calculate interest on the actual opening principal
    interest_due = round(opening_principal * monthly_rate, 2)

    amount_paid = float(row["amount_paid"])

    # Allocate payment to interest first
    interest_paid = min(amount_paid, interest_due)

    # Allocate remaining payment to principal
    principal_paid = min(
        max(amount_paid - interest_paid, 0),
        opening_principal
    )

    unpaid_interest = round(interest_due - interest_paid, 2)

    closing_principal = round(
        opening_principal - principal_paid, 2
    )

    actual_records.append({
        "instalment_number": row["instalment_number"],
        "due_date": row["due_date"],
        "payment_status": row["payment_status"],
        "amount_paid": round(amount_paid, 2),
        "opening_principal": round(opening_principal, 2),
        "interest_due": interest_due,
        "interest_paid": round(interest_paid, 2),
        "principal_paid": round(principal_paid, 2),
        "unpaid_interest": unpaid_interest,
        "closing_principal": closing_principal
    })

    opening_principal = closing_principal

actual_amortisation = pd.DataFrame(actual_records)

display(actual_amortisation)

print(
    "\nActual principal outstanding:",
    round(opening_principal, 2)
)
```

```python

# TARGET: Generate monthly credit snapshots for all 5,000 accounts

# Prepare the required data types
payment_records["due_date"] = pd.to_datetime(
    payment_records["due_date"]
)

payment_records["payment_date"] = pd.to_datetime(
    payment_records["payment_date"]
)

payment_records["amount_due"] = pd.to_numeric(
    payment_records["amount_due"]
)

payment_records["amount_paid"] = pd.to_numeric(
    payment_records["amount_paid"]
).fillna(0)

# Keep the complete list of loan accounts
all_accounts = (
    loan_accounts[["account_id"]]
    .drop_duplicates()
    .copy()
)

portfolio_records = []

# Process each month-end reporting date
for report_date in reporting_dates:

    # Identify payments received by this reporting date
    paid_by_date = (
        payment_records["payment_date"].notna()
        & (payment_records["payment_date"] <= report_date)
    )

    # Calculate outstanding amount as of the reporting date
    outstanding = (
        payment_records["amount_due"]
        - np.where(
            paid_by_date,
            payment_records["amount_paid"],
            0
        )
    ).clip(lower=0)

    # Identify instalments due and still unpaid
    overdue_mask = (
        (payment_records["due_date"] < report_date)
        & (outstanding > 0)
    )

    # Prepare overdue instalment records
    overdue_data = payment_records.loc[
        overdue_mask,
        ["account_id", "due_date"]
    ].copy()

    overdue_data["outstanding_as_of_date"] = (
        outstanding.loc[overdue_mask].to_numpy()
    )

    # Aggregate overdue information account-wise
    account_summary = (
        overdue_data.groupby("account_id")
        .agg(
            overdue_instalments=(
                "outstanding_as_of_date", "size"
            ),
            overdue_exposure=(
                "outstanding_as_of_date", "sum"
            ),
            oldest_overdue_due_date=("due_date", "min")
        )
        .reset_index()
    )

    # Retain accounts with no overdue instalments
    monthly_snapshot = all_accounts.merge(
        account_summary,
        on="account_id",
        how="left",
        validate="one_to_one"
    )

    # Replace missing overdue values with zero
    monthly_snapshot["overdue_instalments"] = (
        monthly_snapshot["overdue_instalments"]
        .fillna(0)
        .astype(int)
    )

    monthly_snapshot["overdue_exposure"] = (
        monthly_snapshot["overdue_exposure"]
        .fillna(0)
        .round(2)
    )

    # Calculate DPD from the oldest overdue instalment
    monthly_snapshot["dpd"] = (
        (report_date - monthly_snapshot["oldest_overdue_due_date"])
        .dt.days
        .fillna(0)
        .astype(int)
    )

    monthly_snapshot["reporting_date"] = report_date

    portfolio_records.append(monthly_snapshot)

# Combine all monthly snapshots
portfolio_snapshots = pd.concat(
    portfolio_records,
    ignore_index=True
)

# Arrange columns
portfolio_snapshots = portfolio_snapshots[
    [
        "account_id",
        "reporting_date",
        "overdue_instalments",
        "overdue_exposure",
        "dpd"
    ]
]

print("Portfolio snapshot generation completed.")
print("Total records:", len(portfolio_snapshots))
print("Unique accounts:", portfolio_snapshots["account_id"].nunique())
print("Reporting dates:", portfolio_snapshots["reporting_date"].nunique())

display(portfolio_snapshots.head(10))
```

```python
# TARGET: Validate all seven CSV headers before PostgreSQL import

from pathlib import Path
import pandas as pd

processed_folder = Path(
    r"C:\DATA ANALYSIS\financial-risk-portfolio-analytics\data\processed"
)

expected_columns = {
    "customer_details": [
        "customer_id", "customer_name", "customer_type",
        "city", "state", "onboarding_date"
    ],

    "product_reference": [
        "product_id", "interest_rate"
    ],

    "loan_accounts": [
        "account_id", "customer_id", "product_id",
        "sanction_amount", "disbursement_date",
        "loan_tenure_months", "account_status", "maturity_date"
    ],

    "repayment_schedule": [
        "account_id", "instalment_number",
        "due_date", "amount_due"
    ],

    "payment_records": [
        "account_id", "instalment_number", "due_date",
        "amount_due", "payment_status", "amount_paid",
        "payment_date", "instalment_outstanding",
        "dpd", "dpd_bucket"
    ],

    "account_summary": [
        "account_id", "overdue_instalments",
        "overdue_exposure", "oldest_overdue_due_date"
    ],

    "portfolio_snapshots": [
        "account_id", "reporting_date",
        "overdue_instalments", "overdue_exposure", "dpd"
    ]
}

validation_results = []

for table_name, expected in expected_columns.items():

    file_path = processed_folder / f"{table_name}.csv"

    actual = pd.read_csv(file_path, nrows=0).columns.tolist()

    validation_results.append({
        "table": table_name,
        "expected_columns": len(expected),
        "actual_columns": len(actual),
        "missing_columns": list(set(expected) - set(actual)),
        "unexpected_columns": list(set(actual) - set(expected)),
        "column_order_matches": actual == expected,
        "status": "PASS" if actual == expected else "FAIL"
    })

header_validation = pd.DataFrame(validation_results)

display(header_validation)

print(
    "\nOverall header validation:",
    "PASSED" if (header_validation["status"] == "PASS").all()
    else "FAILED"
)
```
